```python
import polars as pl
import pandas as pd
```

```python
data = {
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 30, 35],
    "city": ["New York", "Los Angeles", "Chicago"]
    }

polars_df = pd.DataFrame(data)
pandas_df = pl.DataFrame(data)

print("\nPolars DataFrame:")
print(polars_df)

print("\nPandas DataFrame:")
print(pandas_df)
```

```python
import numpy as np
import random
import time
```

```python
#generate a large dataset
large_data = {
    "value 1": np.random.rand(100000000),
    "value 2": np.random.rand(100000000)
    }

polars_large_df = pl.DataFrame(large_data)
pandas_large_df = pd.DataFrame(large_data)

# Measure performance of a simple operation

start_time = time.time()
polar_large_df = polars_large_df.with_columns(
    (pl.col("value 1") * pl.col("value 2")).alias("result")
)
end_time = time.time()
print(f"Polars operation time: {end_time - start_time:.4f} seconds\n")



start_time = time.time()
pandas_large_df["result"] = pandas_large_df["value 1"] * pandas_large_df["value 2"]
end_time = time.time()
print(f"\nPandas operation time: {end_time - start_time:.4f} seconds")
```
