```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
X = iris.data
y = iris.target

pca = PCA(n_components=2)
X_reduced = pca.fit_transform(X)

plt.scatter(X_reduced[:, 0], X_reduced[:, 1], c=y, cmap='viridis')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA of Iris Dataset')
plt.colorbar()
plt.show()
```

```python
cov_matrix = np.cov(X.T)
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print("Eigenvalues:", eigenvalues)
print("Eigenvectors:", eigenvectors)
```

```python
from sklearn.datasets import load_digits

digits = load_digits()
X_digits = digits.data
# y_digits = digits.target

pca_digits = PCA(0.95)
X_digits_projected = pca_digits.fit_transform(X_digits)
X_digits_reconstructed = pca_digits.inverse_transform(X_digits_projected)

fig, ax = plt.subplots(2, 5, figsize=(12, 6))
for i in range(5):
    ax[0, i].imshow(X_digits[i].reshape(8, 8), cmap='gray')
    ax[0, i].set_title('Original Digit')
    ax[1, i].imshow(X_digits_reconstructed[i].reshape(8, 8), cmap='gray')
    ax[1, i].set_title('Reconstructed Digit')
plt.show()
```
