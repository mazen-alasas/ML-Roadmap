```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
data = np.random.normal(loc=50, scale=10, size=500)

plt.figure(figsize=(10, 6))
sns.histplot(data, kde=True, color='blue', bins=20)
plt.title('Histogram with KDE')
plt.xlabel('Value')
plt.ylabel('Density')
plt.grid(True)
plt.show()
```

```python
data = pd.DataFrame({
    'category': ['A', 'B', 'C', 'D'] * 20,
    'value': np.random.randint(10, 50, 80)
})

plt.figure(figsize=(10, 6))
sns.boxplot(x='category', y='value', data=data, palette='Set2')
plt.title('Boxplot with Categorical Data')
plt.xlabel('Category')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```

# violin plot

```python
plt.figure(figsize=(10, 6))
sns.violinplot(x='category', y='value', data=data, palette='Set2')

plt.title('Violin Plot with Categorical Data')
plt.xlabel('Category')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```

# count plot

```python
plt.figure(figsize=(10, 6))

sns.countplot(x='category', data=data, palette='Set2')
plt.title('Count Plot of Categories')   
plt.xlabel('Category')
plt.ylabel('Count')
plt.grid(True)
plt.show()
```

# correlation heatmap

```python
data = pd.DataFrame({
    'x': np.random.rand(100),
    'y': np.random.rand(100),
    'z': np.random.rand(100),
    'i': np.random.rand(100)
})
```

```python
plt.figure(figsize=(10, 8))

sns.heatmap(data.corr(), annot=True, cmap='coolwarm', linewidths=0.7)
plt.title('Heatmap of Correlation Matrix')
plt.show()
```

# regression plot

```python
np.random.seed(42)

x = np.random.rand(100) * 100
y = 2 * x + np.random.normal(0, 25, 100)

plt.figure(figsize=(10, 6))
sns.regplot(x=x, y=y, color='blue', marker='o', line_kws={'color': 'red'})
plt.title('Regression Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

# pair plot

```python
data = sns.load_dataset('iris')

plt.figure(figsize=(10, 8))
sns.pairplot(data, hue='species', palette='Set2')
plt.title('Pair Plot of Iris Dataset')
plt.show()
```
