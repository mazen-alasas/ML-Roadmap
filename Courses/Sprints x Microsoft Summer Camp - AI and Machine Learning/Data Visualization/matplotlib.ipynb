```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# line plot

```python
dates = pd.date_range('2023-01-01', periods=12, freq='M')
sales = [10, 12, 15, 17, 17, 20, 22, 25, 30, 35, 30, 40]

dates, sales
```

```python

plt.figure(figsize=(10, 5))
plt.plot(dates, sales)
plt.xlabel('Date')
plt.ylabel('Sales')
plt.title('Monthly Sales Data')
plt.show()
```

# scatter plot

```python
adversiting_budget = np.random.randint(10, 50, size=20)
sales = adversiting_budget + np.random.randint(5, 10, size=20)

adversiting_budget, sales
```

```python
plt.figure(figsize=(10, 5))
plt.scatter(adversiting_budget, sales)
plt.xlabel('Advertising Budget')
plt.ylabel('Sales')
plt.title('Advertising Budget vs Sales')
plt.show()
```

# bar plot

```python
categories = ['A', 'B', 'C', 'D']
values = [10, 20, 15, 25]

plt.figure(figsize=(10, 5))
plt.bar(categories, values)
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Plot Example')
plt.show()
```

# histograms

```python
np.random.seed(42)
sales = np.random.normal(loc=50, scale=10, size=500)


plt.figure(figsize=(10, 5))
plt.hist(sales, bins=10)
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.title('Histogram of Sales Data')
plt.show()
```

# box plot

```python
data = pd.DataFrame({
    'Category': ['A', 'B', 'C', 'D'] * 10,
    'Values': np.random.randint(10, 40, size=40)
})

data
```

```python
plt.boxplot([data[data['Category'] == cat]['Values'] for cat in data['Category'].unique()],
            labels=data['Category'].unique())
plt.xlabel('Category')
plt.ylabel('Values')
plt.title('Box Plot of Values by Category')
plt.show()
```

# pie plot

```python
labels = ['A', 'B', 'C', 'D']
sizes = [15, 30, 45, 10]

plt.figure(figsize=(10, 5))
plt.pie(sizes, labels=labels)
plt.title('Pie Chart Example')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

# Customizing plot appearance

```python
dates = pd.date_range(start='2023-01-01', periods=12, freq='M')
sales_a = [10, 12, 15, 17, 17, 20, 22, 25, 30, 35, 30, 40]
sales_b = [8, 10, 12, 14, 16, 15, 20, 22, 27, 26, 26, 30]


plt.figure(figsize=(10, 5))
plt.plot(dates, sales_a, label='Product A')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Sales', fontsize=12)
plt.title('Monthly sales of Product A')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(dates, sales_a)
plt.plot(dates, sales_b)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Sales', fontsize=12)
plt.title('Monthly sales of Product A and Product B')
plt.legend(['Product A', 'Product B'])
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(dates, sales_a, color='blue', marker='o', linestyle='-', linewidth=1, markersize=3, label='Product A')
plt.plot(dates, sales_b, color='orange', marker='x', linestyle='--', linewidth=3, markersize=12, label='Product B')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Sales', fontsize=12)
plt.title('Monthly sales of Product A and Product B')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(dates, sales_a, marker='o', linestyle='-', linewidth=1, markersize=3, label='Product A')
plt.plot(dates, sales_b, marker='s', linestyle='--', linewidth=3, markersize=5, label='Product B')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Sales', fontsize=12)
plt.title('Monthly sales of Product A and Product B')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.style.use('dark_background')
plt.show()
```

# Annotating plot

```python
dates = pd.date_range(start='2023-01-01', periods=12, freq='M')
sales_a = [10, 12, 15, 21, 25, 30, 39, 25, 22, 20, 15, 20]
```

```python
plt.style.use('default')

plt.figure(figsize=(10, 5))
plt.plot(dates, sales_a, marker='o', markersize=3)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Sales', fontsize=12)
plt.title('Monthly sales of Product A', fontsize=14)
plt.xticks(rotation=45)
plt.annotate('Highest Sales', xy=(dates[6], sales_a[6]), xytext=(dates[3], sales_a[5] + 5),
             arrowprops=dict(facecolor='black', arrowstyle='->'))
plt.grid(True)
plt.show()
```

# subplots

```python
data = np.linspace(0, 10, 100)

fig, ax = plt.subplots(2, 2, figsize=(10, 10))

ax[0, 0].plot(data, data**2)
ax[0, 0].set_title('y = x^2')

ax[0, 1].plot(data, np.sin(data))
ax[0, 1].set_title('y = sin(x)')

ax[1, 0].plot(data, np.cos(data))
ax[1, 0].set_title('y = cos(x)')

ax[1, 1].plot(data, np.tan(data))
ax[1, 1].set_title('y = tan(x)')

for axes in ax.flat:
    axes.set(xlabel='x', ylabel='y')
    axes.label_outer()  # Hide x labels and tick labels for top plots and y ticks for right plots

plt.tight_layout()
plt.show()
```

# complex layout

```python
dates = pd.date_range(start='2023-01-01', periods=12, freq='M')
sales = [12, 18, 22, 25, 22, 30, 35, 40, 48, 32, 28, 36]
ad_budget = [5, 6, 7, 8, 10, 9, 11, 18, 17, 15, 13, 9]
```

```python
# Create subplots with different sizes
fig = plt.figure(constrained_layout=True, figsize=(14, 10))
gs = fig.add_gridspec(3, 3)

# Plot 1 - Line plot
ax1 = fig.add_subplot(gs[0, :2])
ax1.plot(dates, sales, marker='o', color='b')
ax1.set_title('Sales Over Time')

# Plot 2 - Bar plot
ax2 = fig.add_subplot(gs[0, 2])
ax2.bar(['A', 'B', 'C', 'D'], [20, 35, 30, 35], color='c')
ax2.set_title('Category Comparison')

# Plot 3 - Scatter plot
ax3 = fig.add_subplot(gs[1:, 1:])
ax3.scatter(ad_budget, sales, color='r')
ax3.set_title('Advertising Budget vs Sales')

# Plot 4 - Line plot with markers
ax4 = fig.add_subplot(gs[1:, 0])
ax4.plot(dates, ad_budget, marker='s', linestyle='--', color='g')
ax4.set_title('Advertising Budget Over Time')

# Show the plots
plt.show()
```

# 3d plotting

```python
from mpl_toolkits.mplot3d import Axes3D

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
x, y = np.meshgrid(x, y)
z = np.sin(np.sqrt(x**2 + y**2))
```

```python
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z, cmap='viridis')


ax.set_title('3D Surface Plot')
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
plt.show()
```

```python
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z, cmap='plasma', edgecolor='k', alpha=0.6)
ax.view_init(elev=30, azim=45)

ax.set_title('3D Surface Plot')
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
plt.savefig('3d_surface_plot.png', format='png', dpi=300, bbox_inches='tight')
plt.show()
```
